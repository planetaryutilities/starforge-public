```python
import logging
import time
import textwrap
from IPython.display import Markdown, display
from jl_starforge import update_system_viewer

import scxml4py.helper
from scxmlApp.application import Application
from scxml4py.action import Action
from scxml4py.activity import ThreadedActivity
from scxml4py.event import Event

```

```python

# Installed as status listener, engine looks for "ActionStatusListener" action
class ActionStatusListener(Action):
    # implemented by the developer, stub can be generated
    def __init__(self, theData):
        self.data = theData
        Action.__init__(self, "ActionStatusListener", None, theData)
    
    def execute(self, status):
        print(">>>>ActionStatusListener::execute Status: <" + scxml4py.helper.formatStatus(status) + ">")
        update_system_viewer(scxml_state=scxml4py.helper.formatStatus(status), scxml_content=self.data.getSharedInfo())

class ActionEventListener(Action):
    # implemented by the developer, stub can be generated
    def __init__(self, theData):
        self.data = theData
        Action.__init__(self, "ActionEventListener", None, theData)
    
    def execute(self, theEvent):
        print(">>>>ActionEventListener::execute Event: <" + theEvent.__str__() + ">")


class ActivityOnline(ThreadedActivity):
    # implemented by the developer, stub can be generated
    def __init__(self, theEventQueue, theData):
        ThreadedActivity.__init__(self, "ActivityOnline", theEventQueue, theData)

    def run(self):
        counter = 0
        while self.isRunning() == True:
            logging.getLogger("scxml4py").info("Activity <" + self.getId() + "> is running...")
            print("\nActivityOnline::run <" + self.getId() + "> is running... " + str(counter))
            time.sleep(1)
            counter += 1
            if counter == 3:
                #self.sendInternalEvent(Event("GoOffline"))
                self.setRunning(False)
                break
            
class Data(object):
    # implemented by the developer
    # data shared between actions and activities
    def __init__(self, scxml_doc:str):
        self.mSharedInfo = scxml_doc
    
    def getSharedInfo(self):
        # requires mutex
        return self.mSharedInfo
    
    def setSharedInfo(self, sharedInfo):
        # requires mutex
        self.mSharedInfo = sharedInfo



```

# Display in Jupyter Lab view

```python
simpleSM_scxml='''<?xml version="1.0" ?>
<scxml xmlns="http://www.w3.org/2005/07/scxml" version="1.0" datamodel="ecmascript" initial="OFF">
  <state id="OFF">
    <transition event="Init" target="ONLINE"/>
  </state>
  <state id="READY">
    <transition event="GoOnline" target="ONLINE"/>
  </state>
  <state id="OFFLINE">
    <transition event="GoOnline" target="ONLINE"/>
  </state>
  <state id="ONLINE">
    <invoke id="ActivityOnline"/>
    <transition event="GoOffline" target="OFFLINE"/>
    <transition event="GoReady" target="READY"/>
  </state>
</scxml>'''
```

```python
update_system_viewer(scxml_state="ONLINE", scxml_content=simpleSM_scxml)
```

# Run simulation and animate jupyer lab view

```python
update_system_viewer(scxml_state="ONLINE", scxml_content=simpleSM_scxml)
logging.basicConfig(format='%(asctime)s - %(levelname)s - %(threadName)s - %(module)s - %(funcName)s - %(message)s', level=logging.CRITICAL)
simple_sm_data = Data(simpleSM_scxml)
simple_sm = Application(simpleSM_scxml, actions = [ActionStatusListener, ActionEventListener], activities = [ActivityOnline], data = simple_sm_data)
simple_sm.start()
simple_sm.send_signal("Init", True)
status = simple_sm.get_current_status()
print ("0>>>>>> status = " + status)
print ("#### sending GoOnline")
simple_sm.send_signal("GoOnline", True)
status = simple_sm.get_current_status()
print ("1>>>>>> status = " + status)
time.sleep(2)
print ("#### sending GoOffline")
simple_sm.send_signal("GoOffline", True)
status = simple_sm.get_current_status()
print ("3>>>>>> status = " + status)
time.sleep(2)
print ("#### sending GoOnline")
simple_sm.send_signal("GoOnline", True)
time.sleep(2)
status = simple_sm.get_current_status()
print ("1>>>>>> status = " + status)
print ("#### sending GoReady")
simple_sm.send_signal("GoReady", True)
time.sleep(2)
status = simple_sm.get_current_status()
print ("2>>>>>> status = " + status)
print ("#### sending GoOnline")
simple_sm.send_signal("GoOnline", True)
time.sleep(2)
status = simple_sm.get_current_status()
print ("1>>>>>> status = " + status)
print ("#### sending GoOffline")
simple_sm.send_signal("GoOffline", True)
time.sleep(2)
status = simple_sm.get_current_status()
print ("3>>>>>> status = " + status)
print ("#### sending GoOnline")
simple_sm.send_signal("GoOnline", True)
time.sleep(2)
status = simple_sm.get_current_status()
print ("1>>>>>> status = " + status)
print ("#### sending GoReady")
simple_sm.send_signal("GoReady", True)
time.sleep(2)
status = simple_sm.get_current_status()
print ("2>>>>>> status = " + status)
simple_sm.terminate()
```
